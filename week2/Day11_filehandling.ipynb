## Today concepts
1. create csv,append,read, writefile
2. create txt, append,read,writefile


```python
# open file
import csv
with open('student.csv','r') as stu_file:
    data = csv.reader(stu_file)

    for row in data:
        print(row)
```

```python
with open('student.csv', 'w') as stu_file:
    data = csv.writer(stu_file)
    data.writerow(['111','nermeen','55','AI','wah cantt', '796','85']) 
    print('Data save successfully!')
    
```

```python
with open('student.csv', 'a') as stu_file:
    data = csv.writer(stu_file)
    data.writerow(['112','laiba','55','AI','wah cantt', '796','85']) 
    print('Data save successfully!')
```

```python
with open('student.csv','r',newline='') as stu_file:
    data = csv.reader(stu_file)

    for row in data:
        if row:
            print(row)
```

```python
with open('poem.txt','r') as f:
    data = f.read()

print(data)
```

```python
with open('poem.txt','w') as f:
    f.write('baba baby yes mama ')
    print('Data save successfully!')
```

```python
with open('poem.txt','r') as f:
    data = f.readline()

print(data)
```

```python
with open('poem.txt','r') as f:
    data = f.readlines()

print(data)
```
