```python
# create class
class employee:
    name = 'maraia'
    language = 'Python'
    Department = 'Data analysis'

obj = employee()
print(obj.name, obj.language,obj.Department)
```

```python
# create class with obj prference
class employee:
    name = 'Maria'

obj = employee()
obj.name ='Ali'
print(obj.name)
```

```python
# create class with function
class employee:
    language = 'Python'
    def ger_info(self):
        print(self.language)

obj = employee()
obj.language = 'SQL'
print(obj.language)
```

```python
# create  method with self
class employee:
    def get_info(self,x):
        print(x)

obj = employee()
obj.get_info('python')
```

```python
# create method without self
class employee:
    @staticmethod
    def get_info(name,age):
        print(name)
        print(age)

obj = employee()
obj.get_info('Maria',23)
```

```python
# create constructor
class employee:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def get_info(self):
        print(self.name)
        print(self.age)

obj = employee('maria',23)
obj.get_info()
```

```python
# inheritance concept
class company:
    print('base claas')

class employee(company):
    print('derived class')

obj1=company
obj2= employee
```

```python
# single inheritance
class company:
    print('base claas')

class employee(company):
    print('derived class')


obj1= employee
```

```python
class Employee:
    def show_name(self):
        print("Employee")

class Manager(Employee):
    def show_manager(self):
        print("Manager")


obj = Manager()

obj.show_name()
obj.show_manager()
```

```python
# multilevel inheritance
class company:
    print('base claas')
class employee(company):
    print('derived from company class')
class department(employee):
    print('derived from employee class')

obj1=department
```

```python
class Employee:
    def employee_info(self):
        print("I am an Employee")


class Manager(Employee):
    def manager_info(self):
        print("I am a Manager")


class SeniorManager(Manager):
    def senior_info(self):
        print("I am a Senior Manager")


obj = SeniorManager()

obj.employee_info()
obj.manager_info()
obj.senior_info()
```

```python
# multiple inheritance
class company:
    print('base class one')
class employee:
    print('base class two')
class department(company,employee):
    print('derived class')


obj = department
```

```python
class Employee:
    def work(self):
        print("Employee is working")


class Trainer:
    def teach(self):
        print("Trainer is teaching")


class Manager(Employee, Trainer):
    def manage(self):
        print("Manager is managing")


obj = Manager()

obj.work()
obj.teach()
obj.manage()
```

```python
class Employee:

    def __init__(self, name):
        self.name = name


class Manager(Employee):

    def __init__(self, name, department):
        super().__init__(name)
        self.department = department
        print('hello')


obj = Manager("Maria", "IT")
print(obj.name)
print(obj.department)
```
