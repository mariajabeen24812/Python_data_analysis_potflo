### 🐍 Day 8 — Comprehensions + Lambda

## Today concepts
1. map function
2. filter function
3. sorted function
4. zip function
5. enumerate

```python
# map function
num = [5,8,6,3,9,1]
number = list(map(lambda x : x % 2  == 0, num))
print(number)
```

```python
# map function kar ka har number ka square nakalo
num = [5,8,6,3,9,1]
square = list(map(lambda x :  x * x, num))
print(square)
```

```python
# map function kar ka har number ka square nakalo
num = [5,8,6,3,9,1]
cube = list(map(lambda x :  x * x * x, num))
print(cube)
```

```python
num = [5,8,6,3,9,1]
number = list(filter(lambda x : x % 2  == 0, num))
print(number)
```

```python
# list ma sy greater than 10 number print karo
number = list(filter(lambda x : x > 10 , list(map(int, input('Enter a number:').split() ))))
print(number)
```

```python
# filter() use karke sirf passing marks (>= 50) nikalo.
marks = list(filter(lambda x : x >= 50 , [5,56,15,56,45]))
print(marks)
```

```python
# zip function
name = ['maria', 'ayesha','nermeen']
marks = [85,25,63]
print(list(zip(name,marks)))
```

```python
# zip with dict
names = ["Ali", "Sara", "Maria"]
marks = [80, 90, 85]

students = dict(zip(names, marks))

print(students)
```

```python
item = []
name = ["Ali", "Sara", "Maria"]
items = list(enumerate(name))
print(items)
```

```python
# difference between sort vs sorted
# sort doesnot return list changes ko original ma he karti hai
names = ["Ali", "Sara", "Maria"]
names.sort(reverse= True)
print(names)
# it return list
name = sorted(names,reverse=True)
print(name)
```

```python
# sorted() + lambda use karke students ko marks ke according ascending order mein sort karo.
students = [
    ("Ali", 75),
    ("Sara", 90),
    ("Maria", 85),
    ("Ahmed", 60)
]
students = sorted(students,key=lambda x: x[1],reverse= True)
print(students)
```
