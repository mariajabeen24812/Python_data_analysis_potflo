```python
# User se number lo aur 100 / number calculate karo.
# ZeroDivisionError handle karo.
try:
    num = int(input('Enter a number:'))
    num = 100/num
except ValueError:
    print('value error')
except ZeroDivisionError:
    print('Zero division error')
    
    
```

```python
# else with try except
try:
    num = int(input('Enter a number:'))
    num = 100/num
except ValueError:
    print('value error')
except ZeroDivisionError:
    print('Zero division error')
else:
    print(num)
```

```python
# finally with try except
try:
    num = int(input('Enter a number:'))
    num = 100/num
except ValueError:
    print('value error')
except ZeroDivisionError:
    print('Zero division error')
else:
    print(num)
finally:
    print('You enter invalid value try again')
```

```python
# Exception as e
try:
    num = int(input('Enter a number:'))
    num = 100/num
except Exception as e:
    print('error' , e)
else:
    print(num)
finally:
    print('You enter invalid value try again')
```

```python
# rise with try catch
age = int(input('Enter age:'))
if age < 0:
    raise ValueError('age is not nagative')
```

```python
try:
    age = int(input('enter age:'))
    if age < 0:
        raise ValueError(print('age cannot nagetive'))
    else:
        print(age)
except Exception as e:
    print(e)
```
