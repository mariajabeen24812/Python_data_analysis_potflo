## Today concepts
1. reshapping
2. flattening
3. stacking
4. split

```python
# reshapping converd 1d to 2d and 3d 
import numpy as np
array1 = np.array([1,2,3,4,5,6])
array2 = array1.reshape(2,3)
print(array2)
print(array2.ndim)
```

```python
# flattening convert 2d and 3d to 1d
array1 = np.array([[1,2,3],[5,6,7]])
print(array1.flatten())
```

```python
array1 = np.array([1,2,3,4,5])
array2 = np.array([6,7,8,9,10])
print(np.vstack((array1,array2)))
print(np.hstack((array1,array2)))
```

```python
array1 = np.array([1,2,3,4,5,6])
print(np.split(array1,2))
```
