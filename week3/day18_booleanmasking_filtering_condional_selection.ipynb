```python
# booleanmasking as ka matlab hai true ya false answer da na
import numpy as np
arr = np.array([1,2,3,9,4])
print(arr >= 4)
```

```python
# filtering condition mean ka condition ko filer karna jo as condition ko full fill kara 
arr = np.array([1,2,3,9,4])
filter_values = arr[arr >= 4]
print(filter_values)
```

```python
# conditional selection mean according to conditon value ko select karna 
arr = np.array([3,9])
result= np.where(arr >= 4, 'greater than 3','less than 4')
print(result)
```

```python
# Student Marks Analysis

marks = np.array([35, 78, 45, 90, 28, 65, 82, 39])

# Boolean masking se find karo kaun se students ke marks 50 ya zyada hain.
print(marks >= 50)

# Filtering se sirf passing marks nikalo (passing marks = 50).
result = marks[marks >= 50]
print(result)

# 80 se zyada marks lene wale students find karo.
count = 0 
result = np.where(marks >= 80,'greater than 80','less than 80')
print(result)
```

```python
# Weather Analysis

temperature = np.array([28, 35, 41, 22, 38, 19, 32])

# 35 se zyada temperatures filter karo.
high_tem = temperature[temperature > 35]
print(high_tem)
# 25 se kam temperatures find karo.
low_tem = temperature[temperature < 25]
print(low_tem)
# np.where() use karke 35 ya zyada ko "Hot" aur baqi ko "Normal" label do.
result = np.where(temperature >35, 'Hot','Normal')
print(result)
```

```python
# Monthly Expenses

expenses = np.array([5000, 12000, 3500, 18000, 7000, 22000, 4500])

# 10,000 se zyada expenses filter karo.
high_exp = expenses[expenses > 10000]
print(high_exp)

# 5000 ya kam expenses identify karo.
low_exp = expenses[expenses > 500]
print(low_exp)
# np.where() se 10,000 se zyada expenses ko "Over Budget" aur baqi ko "Within Budget" label do.
result = np.where(expenses > 10000,'Over Budget','Within Budget')
print(result)
```

```python
# City-wise Sales
# Har row ek city ki aur har column ek month ki sales represent karta hai.
sales = np.array([
    [12000, 15000, 9000],
    [8000, 22000, 18000],
    [25000, 30000, 27000],
    [5000, 7000, 6000]
])
# 20,000 ya zyada sales filter karo.
high_sales = sales[sales > 20000]
print(high_sales)
# np.where() se sales 15,000 ya zyada hon to "Good Sales" aur baqi ko "Low Sales" label do.
sales_review = np.where(sales > 15000,'Good Sales','Low Sales')
print(sales_review)
# Sirf third city ki sales mein se 25,000 ya zyada values select karo.
high_sales = sales[2][sales[2] > 20000]
print(high_sales)
```
