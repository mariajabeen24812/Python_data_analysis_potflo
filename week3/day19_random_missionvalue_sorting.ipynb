```python
# find missingvalue, count missing value
arr = np.array([2,5,np.nan,645])
print(arr)
print(np.isnan(arr))
print(np.sum(np.isnan(arr)))

```

```python
# sorting
arr = np.array([11,58,56,765,6,3])
# ascending
print(np.sort(arr))
# descending
print(np.sort(arr)[::-1])
```
