```python
# list comprehension
# List comprehension use karke 1 se 10 tak numbers ki list banao.
# without list comprehension
num = []
for i  in range(1,11):
    num.append(i)
print(num)
```

```python
# with comprehension
num = [i for i in range(1,11)]
print(num)
```

```python
# 1 se 20 tak sirf even numbers ki list banao.
# without list comprehension
num = []
for i in range(1,21):
    if i % 2 == 0:
        num.append(i)

print(num)
```

```python
# with list comprehension
num = [i for i in range(1,21) if i % 2 == 0]
print(num)
```

```python
# number ka square list comprehension se banao.
# without list comprehension
num = []
for i in range(1,21):
        i *=  i
        num.append(i)

print(num)
```

```python
# with comprehension
num = [i*i for i in range(1,21)]
print(num)
```

```python
# sirf 50+ marks ki list banao.
marks = [45, 67, 89, 32, 90, 55]
num = []
for i in marks:
        if i >= 50:
            num.append(i)

print(num)
```

```python
# with comprehension
marks = [45, 67, 89, 32, 90, 55]
num = [i for i in marks if i >= 50]
print(num)
```

```python
# list comprehension se uppercase karo.
names = ["ali", "sara", "maria", "ahmed"]
# with comprehension
num = [i.upper() for i in names]
print(num)
```

```python
# Dictionary comprehension se 1–5 ke numbers aur unke squares ki dictionary banao.
num ={
    i * i 
    for i in range(1,6)
}
print(num)
```

```python
# Dictionary comprehension se sirf 50+ marks wale students nikalo.
marks = {
    "Ali": 45,
    "Sara": 80,
    "Maria": 90,
    "Ahmed": 35
}
student ={
    name : marks
    for name, marks in marks.items() if marks >= 50
}
print(student)
```

```python
# Lambda function banao jo number ka cube return kare
cube = lambda x: x ** 3
cube(2) 
```

```python
# Lambda function banao jo 2 numbers mein se larger number return kare.
check_num = lambda x, y: x if x > y else y
check_num(3,5)
```
