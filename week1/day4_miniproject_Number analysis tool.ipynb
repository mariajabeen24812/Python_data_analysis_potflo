## Number analysis tool


1. Har number print karna hai
2. Even numbers count karne hain
3. Odd numbers count karne hain
4.  Total calculate karna hai
5. Agar user 0 enter kare to break se input lena stop karna hai

```python
total = 0
count_even = 0
count_odd = 0
for i in range(10):
    num = int(input('Enter a number:'))
    if num == 0:
        break

    # find even number 
    if(num % 2 == 0 ):
        count_even += 1
    else:
        count_odd += 1
    
    # calculate total 
    total += num
    


print(f'total even number {count_even}') 
print(f'total odd number {count_odd}') 
print(f'sum of these number {total}') 
```
