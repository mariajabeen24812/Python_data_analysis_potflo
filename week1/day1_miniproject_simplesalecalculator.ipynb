```python
product_name = input('Enter your product_name: ')
pro_qty = int(input('Enter product_quantity:' ))
unit_price = int(input('Enter per unit price : '))
category = input('Enter your produect category :')
discount_prince = float(input('Enter discount prince :'))
total_amount = pro_qty * unit_price
print('PRODUCT SUMMERY')
print(f'Product name : {product_name}')
print(f'Total quantity : {pro_qty}')
print(f'Per unit price : {unit_price}')
print(f'Category:{category}')
print(f'Discount amount : {discount_prince}')
print(f'Total Amount : {total_amount}')
```
