## Operators

```python
# Arithmatic operators
a = 3
b = 9
print('Arithmatic operators')
print(f'Addition {a+b}')
print(f'Subtraction {a-b}')
print(f'Multiplication {a*b}')
print(f'Division {a/b}')
print(f'Floor Division {a//b}')
print(f'Remainder {a%b}')
print(f'power {a**b}')
```

```python
# Assignment operators
print('\n Assignment operator ')
x = 50
print(x)
x += 20
print(x)
x -= 30
print(x)
x *= 2
print(x)
x /= 5
print(x)
x %= 5
print(x)
x //= 2
print(x)
x **= 3
print(x)
```

```python
#Comparison operator
print(' Comparison operator')
age = int(input('Enter your age : '))
print(f'check age is greater than 18 than eldel {age >= 18}  ')
print(f'check age is less than 18 than younger {age <= 18}  ')
```

```python
# logical operator
print('Logical operators')

age = int(input('Enter your age: '))

# AND operator
print(f'Check age is between 18 and 40: {age >= 18 and age <= 40}')

# OR operator
print(f'Check age is less than 18 OR greater than 40: {age < 18 or age > 40}')

# NOT operator
print(f'Check age is NOT between 18 and 40: {not(age >= 18 and age <= 40)}')
```

## String and their slicing and function

```python
# string basic
# create a variable and  store a name
name = 'Ayan'
print(f'name : {name}')
# Q2: Ek sentence store karo aur uski length find karo.
sentence = 'My name is Ayan'
print(sentence,'\nlength of the sentence ' , len(sentence))
# Q3: input a name and print name with hello
name = input('Enter your name :')
print(f'Hello {name}')
```

```python
# string indexing
# Q4: String ka first character print karo.
print('first character of name', name[0])
# Q5: String ka last character print karo.
text = "Python"
print(f'last character of strint {text[5]}')
# Q6: String ka third character print karo.
text = "Data Analysis"
print(f'third character of strint {text[7]}')
```

```python
# string slicing
text = "Data Analysis"
# Q5: string ka first four character print karo
print(f'first four character of text {text[ 0: 4]}')
print(f'first four character or text {text[ : 4]}')
print(f'text character  {text[ -13: -1]}')
```

```python
# string functions
text = "  Data Analysis"
# 1-len()
print(len(text))
# 2- endswith
print(text.endswith('sis'))
# 3- startswith
print(text.startswith('da'))
# 4- strip
print(text.strip())
# 5- upper
print(text.upper())
# 6- lower
print(text.lower())
# 7- capitalize
print(text.capitalize())
# 8- replace
print(text.replace('Analysis','science'))
# 9- split
print(text.split())
# 10- join
list  = ('maria','jabeen')
print(" ".join(list))
# 11- find
print(text.find('D'))
# 12- count
print(text.count('a'))
# 13- isdigit()
print(text.isdigit())
# 14- isnumeric()
print(text.isnumeric())
# 15- isalpha()
print(text.isalpha())
# 16-title()
print(text.title())
```
