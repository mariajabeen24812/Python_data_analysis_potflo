## Text Analyzer

```python
# what do in this mini project

# User se ek sentence input lena hai, phir program ye cheezen show kare:

    # Original text
    # Text ki total length → len()
    # Text uppercase → upper()
    # Text lowercase → lower()
    # Text title case → title()
    # Extra spaces remove → strip()
    # First character → indexing
    # Last character → negative indexing
    # First 5 characters → slicing
    # Reverse text → slicing
    # Total words → split()
    # "a" kitni dafa hai → count()
    # "Python" text mein hai ya nahi → find() ya in
    # Text sirf alphabets par based hai ya nahi → isalpha()
```

```python
sentence = input('Enter a sentence : ')
print(f'Original text {sentence}')
print(f'total sentence length {len(sentence)}')
print(f'sentence in upper case{sentence.upper()} ')
print(f'sentence in lower case{sentence.lower()} ')
print(f'sentence in title case{sentence.title()} ')
print(f'remove extra space in sentence {sentence.strip()} ')
print(f'sentence  first character{sentence[4]} ') # first character is space in string
print(f'sentence  last character{sentence[-1]} ')
print(f'first five character{sentence[0:6]}')
print(f'Reverse text{sentence[::-1 ]}')
print(f'convert sentence in list{sentence.split()}')
print(f'count a in sentence which is {sentence.count('a')}')
print(f'find python in sentence {sentence.find('python')}')
print(f'check karo ka sentence ma sarf alphabets hai ya nahe :  {sentence.isalpha()}')

```
