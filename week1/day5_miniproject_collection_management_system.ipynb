## Collection management system


```python
# Part 1 — Collection Create Karna
item = input('Enter a item:').split()
print(item)
print(type(item))
```

```python
# access first,third last item in the collection
print(f'first item is {item[0]}')
print(f'last item is {item[-1]}')
print(f'third item is {item[2]}')
```

```python
# add new item 
item.append('pencil')
print(item)
# Collection mein ek new item second position par add karo
item.insert(1,'Camera')
print(item)
```

```python
# User se ek item ka naam lo aur collection se remove karo
item = ['Laptop', 'Camera', 'Mobile', 'Book', 'Headphones', 'Charger', 'Notebook', 'Pen', 'Bag', 'pencil']
remove_item = input('Enter an item that remove from collection:')
remove_item = remove_item.title()
if(remove_item not in item):
    print('item not found')
else:
    item.remove(remove_item)    
    print(item)
# Collection ka last item remove karo
item.pop()
print(item)

```

```python
# Collection mein total items count karo.
print('total item in list',len(item))
```

```python
# User se item ka naam lo aur check karo ke woh collection mein hai ya nahi.
item = ['Laptop', 'Camera', 'Mobile', 'Book', 'Headphones', 'Charger', 'Notebook', 'Pen', 'Bag', 'pencil']
check_item = input('Enter an item that check in collection:')
check_item = check_item.title()
if check_item in item:
    print('is avalible')
else:
    print('is not avalible')
```

```python
# Agar same item multiple times collection mein ho, check karo woh kitni baar hai
print('count laptop in list',item.count('Laptop'))
```

```python
# User se item lo aur uska index find karo.
check_item = input('Enter an item that check in collection:')
check_item = check_item.title()
print('item in index',item.index(check_item))
```

```python
# Collection ko alphabetical order mein arrange karo.
item.sort()
print(item)
item.sort(reverse=True)
print(item)
```
