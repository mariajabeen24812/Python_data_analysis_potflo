```python
# User se age input lo.
# Agar age >= 18 ho to "Eligible"
# warna "Not Eligible" print karo
age = int(input('Enter your age: '))
print('Eligible criteria is 18 so ')
if age >= 18:
    print(f'you are Eligible because your age is {age}')
else:
    print(f'you are NOT Eligible because your age is {age}')
```

```python
# Marks ke according grade print karo:
#
# 90 or above  = A+
# 80-89        = A
# 70-79        = B
# 60-69        = C
# below 60     = Fail
marks= int(input('Enter your marks: '))
print(f'your marks is {marks}')
if marks >= 90:
    print('Grade : A+')
elif(marks <= 89 and marks >= 80 ):
    print('Grade : A')
elif(marks <= 79 and marks >= 70 ):
    print('Grade : B')
elif(marks <= 69 and marks >= 60 ):
    print('Grade : C')
elif(marks <= 59 and marks >= 50 ):
    print('Grade : D')
elif(marks <= 49 and marks >= 40 ):
    print('Grade : E')
else:
    print('you are fail')
```

```python
# User se age input lo.
#
# 0-12   = Child
# 13-17  = Teenager
# 18-59  = Adult
# 60+    = Senior
age = int(input('Enter your age: '))

if age <= 12:
    print(f'you are Child because your age is {age}')
elif(age >= 13 and age <=17):
    print(f'you are Teenager because your age is {age}')
elif(age >=18  and age <= 59):
    print(f'you are Adult because your age is {age}')
else:
    print(f'you are Senior because your age is {age}')
```
