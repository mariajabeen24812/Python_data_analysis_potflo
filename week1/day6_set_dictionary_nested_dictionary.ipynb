```python
# create a dictionary
stu = {
    'name' : 'maria',
    'roll_no' : 33,
    "total_number" : 789
}
print(stu)
```

```python
# access dictionary by it key
print(stu['total_number'])
```

```python
# add new key value
stu.update({'total_subject': 5})
print(stu)
```

```python
# moditfy value
stu['name'] = 'Ali'
print(stu)
```

```python
# remove value in dictionary
stu = {'name': 'Ali', 'roll_no': 33, 'total_number': 789, 'total_subject': 5}
stu.pop('name') 
print(stu)
# remove only last item
stu.popitem()
print(stu)
# it remove whole dictionary
stu.clear()
print(stu)
```

```python
# find all keys
stu = {'name': 'Ali', 'roll_no': 33, 'total_number': 789, 'total_subject': 5}
print(stu.keys())
# find all value
print(stu.values())
# find both keys and values
print(stu.items())
```

```python
# with and without using function we access the value in the dictionary
print(stu.get('name'))
print(stu['name'])
```

```python
# access keys using loop
stu = {'name': 'Ali', 'roll_no': 33, 'total_number': 789, 'total_subject': 5}
for key in stu.keys():
    print(key)
```

```python
# access values using loop
stu = {'name': 'Ali', 'roll_no': 33, 'total_number': 789, 'total_subject': 5}
for value in stu.values():
    print(value)
```

```python
# access both using loop
stu = {'name': 'Ali', 'roll_no': 33, 'total_number': 789, 'total_subject': 5}
for key, values in stu.items():
    print( key ," : ", values)
```

```python
# Nested dictionary
students = {
    'student1': {
        'name': 'Ali',
        'age': 20,
        'marks': 85
    },

    'student2': {
        'name': 'Sara',
        'age': 21,
        'marks': 90
    }
}
# access name in bohat dictioneries
print(students['student1']['name'])
print(students['student2']['name'])
# add new key and value 
students['student1'].update({'total_marks':'487'})
print(students['student1'])
```

```python
students = {
    'student_name': {
      'stu1_name' : 'Maria',
      'stu2_name' : 'Nermeen',
      'stu3_name' :  'Ayesha',
      'stu4_name' :  'Mehnoor'
    },

    'student_marks': {
       'stu1_marks' : 34,
       'stu2_marks' : 345,
       'stu3_marks' : 234,
       'stu4_marks' : 566
    }
}
print(students['student_name']['stu1_name']," : ",students['student_marks']['stu1_marks'])
```

```python
# create set
name = {'maria','ayesha','nermeen','mehnoor','ali'}
print(name)
print(type(name))
```

```python
# remove douplicate  values
name = {'maria','ayesha','nermeen','mehnoor','ali','maria','ali'}
print(name)

print(type(name))
```

```python
# add value
name.add('zara')
print(name)
# delete value with error if not found
name.remove('ali')
print(name)
# delete value without error if not found
name.discard('ayesha')
print(name)
```

```python
name = {'ayesha', 'mehnoor', 'nermeen', 'ali', 'maria', 'zara'}
marks = {12,34,56,23,78,54}
print(name.union(marks))
print(name.intersection(marks))
print(name.difference(marks))
print(name.issubset(marks))
print(name.issuperset(marks))
```
