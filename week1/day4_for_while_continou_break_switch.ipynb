```python
game = "Football"

match game:
    case "Football":
        print("You selected Football")
    case "Cricket":
        print("You selected Cricket")
    case "Hockey":
        print("You selected Hockey")
    case _:
        print("Invalid game")
```

```python
# for loop use karke 1 se 10 tak numbers print karo.
for i in range(11):
    print(i)
# i = 0
# for i <=10:
#     i+=1

```

```python
# range() use karke 2, 4, 6, 8, 10 print karo.
for i in range(2,11,2):
    print(i)
```

```python
# for loop se 10 se 1 tak reverse counting print karo
for i in range(10,0,-1):
    print(i)
```

```python
# 1 se 20 tak sirf even numbers print karo.
for i in range(2,21,2):
    print(i)
```

```python
# 1 se 20 tak sirf odd numbers print karo.
for i in range(1,20,2):
    print(i)
```

```python
# while loop use karke 1 se 10 tak numbers print karo
i=0
while i <=10 :
    print(i)
    i+=1 
```

```python
# while loop use karke 10 se 1 tak reverse counting karo
i =10 
while i >= 1 :
    print(i)
    i-=1
```

```python
# for loop mein break use karo aur 1 se 10 tak numbers print karte hue 5 par loop stop karo
i=0
while i <=10 :
    print(i)
    if i == 5:
        break
    i+=1
```

```python
# for loop mein continue use karo aur 1 se 10 tak numbers print karo, lekin 5 ko skip karo
for i in range(11):
   if i == 5:
        continue
   else:
       print(i)
        
```

```python
# User se ek number lo aur us number ka multiplication table 1 se 10 tak print karo.
number = int(input('Enter your number:'))
for i in range(1,11):
    print(f' {number} * {i} = {number * i }')
```
