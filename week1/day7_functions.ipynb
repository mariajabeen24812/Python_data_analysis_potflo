```python
# calculator(a, b, operation) function banao.
def calculator(a,b,operation):
    if operation == '+':
        return a + b
    elif operation == '-':
        return a - b
    elif operation == '*':
        return a * b
    elif operation == '/':
        return a / b
    elif operation == '%':
        return a % b
    else:
        print('invalid operation')


x = int(input('Enter a number'))
y = int(input('Enter a number'))
operator = input('Enter an operation')

print(calculator(x,y,operator))
```

```python
# get_grade(marks) function banao.
def get_grade(mark):
    if mark >= 90 and mark <=100 :
        print('A+')
    elif marks >= 80 and marks <= 89:
        print('A')
    elif marks >= 70 and marks <= 79:
        print('B')  
    
    elif marks >= 60 and marks <= 69:
        print('c')  
    elif marks >= 50 and marks <= 59:
        print('D') 
    else:
        print('F') 
marks = int(input('Enter a marks: '))
get_grade(marks)
```

```python
# largest_number(numbers)
largest = 0
def largest_number(no):
    first = no[0]
    for i in no:
        if i >= first:
            largest = i 
    return largest

num =list(map(int,input('Enter a number:').split()))
largest_number(num)
```

```python
# count_even_odd(numbers) function banao
def count_even_oddnumber(num):
    counteven = 0
    countodd = 0
    for i in num:
        if i % 2 == 0:
            counteven = counteven + 1
            print(f'{i} is even')
        else:
            countodd = countodd + 1
            print(f'{i} is odd')
    print(f'Total even number in list {counteven}')
    print(f'Total odd number in list {countodd}')
number =list(map(int,input('Enter a number:').split()))
count_even_oddnumber(number)
```

```python
# check_eligibility(age, marks, attendance) elifible if
# Age >= 18
# Marks >= 50
# Attendance >= 75
def check_eligibility(age,marks,attendance):
    if age >= 18 and marks >= 50 and attendance >=75:
        print('you are eligible')
    else:
        print('you are not eligible')

age = int(input('Enter your age: '))
marks = int(input('Enter your marks'))
attendance = int(input('Enter your attendance'))
check_eligibility(age,marks,attendance)
```
