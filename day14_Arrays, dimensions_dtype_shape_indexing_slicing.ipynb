## Today concepts
1. 1D array,2D array, 3D array
2. shpes
3. dimention
4. dtype
5. indexing

```python
# create 1d array
import numpy as np
array1 = np.array([1,2,3,4])
print(array1)
# find dimention
print(array1.ndim)
# find shape
print(array1.shape)
# find dtype
print(array1.dtype)
# use indexing
print(array1[:2])
```

```python
# creaate 2d array
array2 = np.array([[2.3,2,3],[4,3,5],[4,0,7]])
print(array2)
# find dimention
print(array2.ndim)
# find shape
print(array2.shape)
# find dtype
print(array2.dtype)
# use indexing
print(array2[1][2])
```

```python
# create 3d array
array3 = np.array([[[3.3,0,3],[4,3,5]],[[4,0,7],[4,6,6.7]]])
print(array3)
# find dimention
print(array3.ndim)
# find shape (layer,row and column)
print(array3.shape)
# find dtype
print(array3.dtype)
# use indexing
print(array3[1][1][2])
```
